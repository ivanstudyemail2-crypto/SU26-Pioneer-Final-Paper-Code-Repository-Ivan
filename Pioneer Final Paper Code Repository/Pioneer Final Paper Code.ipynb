```python
import numpy as np
import matplotlib.pyplot as plt

from qiskit import QuantumCircuit, transpile
from qiskit.quantum_info import Statevector, Operator, state_fidelity

from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, coherent_unitary_error
```

```python
theta = 1.1
phi = 0.7

input_circuit = QuantumCircuit(1)
input_circuit.ry(theta, 0)
input_circuit.rz(phi, 0)

target_state = Statevector.from_instruction(input_circuit)

print(target_state)




def build_normal_teleportation(theta=1.1, phi=0.7):
    qc = QuantumCircuit(3)

    # Prepare the state to be teleported on q0
    qc.ry(theta, 0)
    qc.rz(phi, 0)

    # Create Bell pair between q1 and q2
    qc.h(1)
    qc.cx(1, 2)

    qc.barrier()

    # Bell-basis operation
    qc.cx(0, 1)
    qc.h(0)

    qc.barrier()

    # Coherent version of teleportation corrections
    qc.cx(1, 2)
    qc.cz(0, 2)

    # Save ONLY Bob's output qubit
    qc.save_density_matrix(
        qubits=[2],
        label="rho_out"
    )

    return qc




PAULIS = ["I", "X", "Y", "Z"]

pauli_to_bits = {
    "I": (0, 0),
    "X": (1, 0),
    "Y": (1, 1),
    "Z": (0, 1)
}

bits_to_pauli = {
    (0, 0): "I",
    (1, 0): "X",
    (1, 1): "Y",
    (0, 1): "Z"
}


def apply_pauli(qc, p, qubit):
    if p == "X":
        qc.x(qubit)
    elif p == "Y":
        qc.y(qubit)
    elif p == "Z":
        qc.z(qubit)
    # I means do nothing




def cx_conjugated_paulis(pc, pt):
    xc, zc = pauli_to_bits[pc]
    xt, zt = pauli_to_bits[pt]

    xc2 = xc
    zc2 = zc ^ zt

    xt2 = xt ^ xc
    zt2 = zt

    return (
        bits_to_pauli[(xc2, zc2)],
        bits_to_pauli[(xt2, zt2)]
    )




def cz_conjugated_paulis(pc, pt):
    xc, zc = pauli_to_bits[pc]
    xt, zt = pauli_to_bits[pt]

    xc2 = xc
    zc2 = zc ^ xt

    xt2 = xt
    zt2 = zt ^ xc

    return (
        bits_to_pauli[(xc2, zc2)],
        bits_to_pauli[(xt2, zt2)]
    )




def twirled_cx(qc, control, target, rng):
    pc = rng.choice(PAULIS)
    pt = rng.choice(PAULIS)

    # Pauli before CNOT
    apply_pauli(qc, pc, control)
    apply_pauli(qc, pt, target)

    qc.cx(control, target)

    # Compensating Pauli after CNOT
    pc_after, pt_after = cx_conjugated_paulis(pc, pt)

    apply_pauli(qc, pc_after, control)
    apply_pauli(qc, pt_after, target)


def twirled_cz(qc, control, target, rng):
    pc = rng.choice(PAULIS)
    pt = rng.choice(PAULIS)

    apply_pauli(qc, pc, control)
    apply_pauli(qc, pt, target)

    qc.cz(control, target)

    pc_after, pt_after = cz_conjugated_paulis(pc, pt)

    apply_pauli(qc, pc_after, control)
    apply_pauli(qc, pt_after, target)




def build_twirled_teleportation(rng, theta=1.1, phi=0.7):
    qc = QuantumCircuit(3)

    # Prepare state
    qc.ry(theta, 0)
    qc.rz(phi, 0)

    # Bell pair
    qc.h(1)
    twirled_cx(qc, 1, 2, rng)

    qc.barrier()

    # Bell-basis operation
    twirled_cx(qc, 0, 1, rng)
    qc.h(0)

    qc.barrier()

    # Corrections
    twirled_cx(qc, 1, 2, rng)
    twirled_cz(qc, 0, 2, rng)

    qc.save_density_matrix(
        qubits=[2],
        label="rho_out"
    )

    return qc




rng = np.random.default_rng(1234)

twirled = build_twirled_teleportation(rng)

twirled.draw("mpl")




def make_coherent_noise_model(epsilon):
    noise_model = NoiseModel()

    # Two-qubit error circuit
    error_circuit = QuantumCircuit(2)

    # Systematic unwanted rotation on the second qubit
    error_circuit.rx(epsilon, 1)

    error_unitary = Operator(error_circuit).data

    coherent_error = coherent_unitary_error(error_unitary)

    # Apply after every CX and CZ
    noise_model.add_all_qubit_quantum_error(
        coherent_error,
        ["cx", "cz"]
    )

    return noise_model




def calculate_fidelity(circuit, noise_model, target_state):
    simulator = AerSimulator(
        method="density_matrix",
        noise_model=noise_model
    )

    transpiled = transpile(
        circuit,
        simulator,
        optimization_level=0
    )

    result = simulator.run(transpiled).result()

    rho_out = result.data(0)["rho_out"]

    fidelity = state_fidelity(target_state, rho_out)

    return float(fidelity)




normal = build_normal_teleportation(theta, phi)

no_noise = make_coherent_noise_model(0.0)

F = calculate_fidelity(
    normal,
    no_noise,
    target_state
)

print("Ideal fidelity:", F)




rng = np.random.default_rng(1234)

for i in range(10):
    circuit = build_twirled_teleportation(rng, theta, phi)

    F = calculate_fidelity(
        circuit,
        no_noise,
        target_state
    )

    print(i, F)




#Actual Experiment
noise_strengths = np.linspace(0.0, 0.30, 13)

N_TWIRLS = 100

normal_fidelities = []
twirled_mean_fidelities = []
twirled_std_fidelities = []
twirled_sem_fidelities = []

rng = np.random.default_rng(2026)

for epsilon in noise_strengths:

    noise_model = make_coherent_noise_model(epsilon)

    # -------------------------
    # No Pauli twirling
    # -------------------------
    normal_circuit = build_normal_teleportation(theta, phi)

    F_normal = calculate_fidelity(
        normal_circuit,
        noise_model,
        target_state
    )

    normal_fidelities.append(F_normal)

    # -------------------------
    # Pauli twirling
    # -------------------------
    twirled_fidelities = []

    for _ in range(N_TWIRLS):

        twirled_circuit = build_twirled_teleportation(
            rng,
            theta,
            phi
        )

        F = calculate_fidelity(
            twirled_circuit,
            noise_model,
            target_state
        )

        twirled_fidelities.append(F)

    mean_F = np.mean(twirled_fidelities)
    std_F = np.std(twirled_fidelities, ddof=1)
    sem_F = std_F / np.sqrt(N_TWIRLS)

    twirled_mean_fidelities.append(mean_F)
    twirled_std_fidelities.append(std_F)
    twirled_sem_fidelities.append(sem_F)

    print(
        f"epsilon={epsilon:.3f} | "
        f"No PT={F_normal:.5f} | "
        f"PT={mean_F:.5f}"
    )




plt.figure(figsize=(9, 6))

plt.plot(
    noise_strengths,
    normal_fidelities,
    marker="o",
    label="Without Pauli Twirling"
)

plt.errorbar(
    noise_strengths,
    twirled_mean_fidelities,
    yerr=twirled_sem_fidelities,
    marker="o",
    capsize=4,
    label="With Pauli Twirling"
)

plt.xlabel("Coherent Error Strength ε (radians)")
plt.ylabel("Teleportation Fidelity")
plt.title(
    "Effect of Pauli Twirling on Quantum Teleportation Fidelity"
)

plt.ylim(0, 1.02)

plt.grid(alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# PART 1:
# 3-QUBIT TELEPORTATION WITH MULTIPLE INPUT STATES
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from qiskit import QuantumCircuit, transpile
from qiskit.quantum_info import Statevector, Operator, state_fidelity

from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, coherent_unitary_error


# ============================================================
# 1. DEFINE INPUT STATES
# ============================================================

TEST_STATES = ["0", "1", "+", "+i", "arbitrary"]


def prepare_state(qc, qubit, state_name):

    if state_name == "0":
        # Qubit already starts in |0>
        pass

    elif state_name == "1":
        # X|0> = |1>
        qc.x(qubit)

    elif state_name == "+":
        # H|0> = |+>
        qc.h(qubit)

    elif state_name == "+i":
        # H then S creates |+i>
        qc.h(qubit)
        qc.s(qubit)

    elif state_name == "arbitrary":
        # Your original state
        qc.ry(1.1, qubit)
        qc.rz(0.7, qubit)

    else:
        raise ValueError("Unknown state")


def get_target_state(state_name):

    qc = QuantumCircuit(1)

    prepare_state(
        qc,
        0,
        state_name
    )

    return Statevector.from_instruction(qc)


# ============================================================
# 2. PAULI TWIRLING HELPER FUNCTIONS
# ============================================================

PAULIS = ["I", "X", "Y", "Z"]

pauli_to_bits = {
    "I": (0, 0),
    "X": (1, 0),
    "Y": (1, 1),
    "Z": (0, 1)
}

bits_to_pauli = {
    (0, 0): "I",
    (1, 0): "X",
    (1, 1): "Y",
    (0, 1): "Z"
}


def apply_pauli(qc, p, qubit):

    if p == "X":
        qc.x(qubit)

    elif p == "Y":
        qc.y(qubit)

    elif p == "Z":
        qc.z(qubit)


def cx_conjugated_paulis(pc, pt):

    xc, zc = pauli_to_bits[pc]
    xt, zt = pauli_to_bits[pt]

    xc2 = xc
    zc2 = zc ^ zt

    xt2 = xt ^ xc
    zt2 = zt

    return (
        bits_to_pauli[(xc2, zc2)],
        bits_to_pauli[(xt2, zt2)]
    )


def cz_conjugated_paulis(pc, pt):

    xc, zc = pauli_to_bits[pc]
    xt, zt = pauli_to_bits[pt]

    xc2 = xc
    zc2 = zc ^ xt

    xt2 = xt
    zt2 = zt ^ xc

    return (
        bits_to_pauli[(xc2, zc2)],
        bits_to_pauli[(xt2, zt2)]
    )


def twirled_cx(qc, control, target, rng):

    # Random Pauli gates
    pc = rng.choice(PAULIS)
    pt = rng.choice(PAULIS)

    # Before CX
    apply_pauli(qc, pc, control)
    apply_pauli(qc, pt, target)

    # Actual CX
    qc.cx(control, target)

    # Calculate compensating Paulis
    pc_after, pt_after = cx_conjugated_paulis(
        pc,
        pt
    )

    # After CX
    apply_pauli(qc, pc_after, control)
    apply_pauli(qc, pt_after, target)


def twirled_cz(qc, control, target, rng):

    pc = rng.choice(PAULIS)
    pt = rng.choice(PAULIS)

    apply_pauli(qc, pc, control)
    apply_pauli(qc, pt, target)

    qc.cz(control, target)

    pc_after, pt_after = cz_conjugated_paulis(
        pc,
        pt
    )

    apply_pauli(qc, pc_after, control)
    apply_pauli(qc, pt_after, target)


# ============================================================
# 3. NORMAL 3-QUBIT TELEPORTATION
# ============================================================

def build_normal_3q(state_name):

    qc = QuantumCircuit(3)

    # State to teleport is on q0
    prepare_state(qc, 0, state_name)

    # Create Bell pair q1-q2
    qc.h(1)
    qc.cx(1, 2)

    qc.barrier()

    # Bell-basis operations
    qc.cx(0, 1)
    qc.h(0)

    qc.barrier()

    # Coherent corrections
    qc.cx(1, 2)
    qc.cz(0, 2)

    # q2 is Bob's final output
    qc.save_density_matrix(
        qubits=[2],
        label="rho_out"
    )

    return qc


# ============================================================
# 4. TWIRLED 3-QUBIT TELEPORTATION
# ============================================================

def build_twirled_3q(rng, state_name):

    qc = QuantumCircuit(3)

    prepare_state(qc, 0, state_name)

    # Bell pair
    qc.h(1)
    twirled_cx(qc, 1, 2, rng)

    qc.barrier()

    # Bell-basis operations
    twirled_cx(qc, 0, 1, rng)
    qc.h(0)

    qc.barrier()

    # Corrections
    twirled_cx(qc, 1, 2, rng)
    twirled_cz(qc, 0, 2, rng)

    qc.save_density_matrix(
        qubits=[2],
        label="rho_out"
    )

    return qc


# ============================================================
# 5. COHERENT NOISE MODEL
# ============================================================

def make_coherent_noise_model(epsilon):

    noise_model = NoiseModel()

    error_circuit = QuantumCircuit(2)

    # Systematic unwanted Rx rotation
    # on second qubit of every two-qubit gate
    error_circuit.rx(epsilon, 1)

    error_unitary = Operator(
        error_circuit
    ).data

    coherent_error = coherent_unitary_error(
        error_unitary
    )

    noise_model.add_all_qubit_quantum_error(
        coherent_error,
        ["cx", "cz"]
    )

    return noise_model


# ============================================================
# 6. FIDELITY FUNCTION
# ============================================================

def calculate_fidelity(
    circuit,
    noise_model,
    target_state
):

    simulator = AerSimulator(
        method="density_matrix",
        noise_model=noise_model
    )

    transpiled = transpile(
        circuit,
        simulator,
        optimization_level=0,
        seed_transpiler=1234
    )

    result = simulator.run(
        transpiled
    ).result()

    rho_out = result.data(0)["rho_out"]

    F = state_fidelity(
        target_state,
        rho_out
    )

    return np.clip(
        float(np.real(F)),
        0,
        1
    )


# ============================================================
# 7. ZERO-NOISE CHECK
# ============================================================

print("ZERO-NOISE CHECK\n")

no_noise = make_coherent_noise_model(0.0)

for state_name in TEST_STATES:

    target = get_target_state(state_name)

    circuit = build_normal_3q(
        state_name
    )

    F = calculate_fidelity(
        circuit,
        no_noise,
        target
    )

    print(
        f"{state_name}: {F:.10f}"
    )


# ============================================================
# 8. ACTUAL EXPERIMENT
# ============================================================

noise_strengths = np.linspace(
    0,
    0.30,
    13
)

N_TWIRLS = 100

rows = []


for state_name in TEST_STATES:

    print(
        "\n===================="
    )

    print(
        "STATE:",
        state_name
    )

    print(
        "===================="
    )

    target_state = get_target_state(
        state_name
    )

    for epsilon in noise_strengths:

        noise_model = make_coherent_noise_model(
            epsilon
        )

        # ------------------------------------
        # WITHOUT PAULI TWIRLING
        # ------------------------------------

        normal_circuit = build_normal_3q(
            state_name
        )

        F_normal = calculate_fidelity(
            normal_circuit,
            noise_model,
            target_state
        )

        # ------------------------------------
        # WITH PAULI TWIRLING
        # ------------------------------------

        twirled_fidelities = []

        # Same random sample structure
        # at each epsilon
        rng = np.random.default_rng(2026)

        for _ in range(N_TWIRLS):

            twirled_circuit = build_twirled_3q(
                rng,
                state_name
            )

            F = calculate_fidelity(
                twirled_circuit,
                noise_model,
                target_state
            )

            twirled_fidelities.append(F)

        PT_mean = np.mean(
            twirled_fidelities
        )

        PT_std = np.std(
            twirled_fidelities,
            ddof=1
        )

        PT_sem = PT_std / np.sqrt(
            N_TWIRLS
        )

        # Save results
        rows.append({
            "state": state_name,
            "epsilon": epsilon,
            "no_PT": F_normal,
            "PT_mean": PT_mean,
            "PT_std": PT_std,
            "PT_sem": PT_sem,
            "delta_F": PT_mean - F_normal
        })

        print(
            f"epsilon={epsilon:.3f} | "
            f"No PT={F_normal:.5f} | "
            f"PT={PT_mean:.5f}"
        )


# ============================================================
# 9. PUT RESULTS IN DATAFRAME
# ============================================================

results_3q = pd.DataFrame(rows)

print("\nCOMPLETE 3-QUBIT DATA:")
print(results_3q)


# Optional: save it
results_3q.to_csv(
    "3qubit_multiple_states.csv",
    index=False
)


# ============================================================
# 10. GRAPH EACH STATE
# ============================================================

for state_name in TEST_STATES:

    data = results_3q[
        results_3q["state"] == state_name
    ]

    plt.figure(figsize=(9, 6))

    plt.plot(
        data["epsilon"],
        data["no_PT"],
        marker="o",
        label="Without Pauli Twirling"
    )

    plt.errorbar(
        data["epsilon"],
        data["PT_mean"],
        yerr=data["PT_sem"],
        marker="o",
        capsize=4,
        label="With Pauli Twirling"
    )

    plt.xlabel(
        "Coherent Error Strength ε (radians)"
    )

    plt.ylabel(
        "Teleportation Fidelity"
    )

    plt.title(
        f"3-Qubit Teleportation — Input {state_name}"
    )

    plt.ylim(0, 1.02)

    plt.grid(alpha=0.3)

    plt.legend()

    plt.tight_layout()

    plt.show()
```

```python
# ============================================================
# PART 2:
# 5-QUBIT / TWO-STAGE TELEPORTATION
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from qiskit import QuantumCircuit, transpile
from qiskit.quantum_info import Statevector, Operator, state_fidelity

from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, coherent_unitary_error


# ============================================================
# 1. INPUT STATES
# ============================================================

TEST_STATES = ["0", "1", "+", "+i", "arbitrary"]


def prepare_state(qc, qubit, state_name):

    if state_name == "0":
        pass

    elif state_name == "1":
        qc.x(qubit)

    elif state_name == "+":
        qc.h(qubit)

    elif state_name == "+i":
        qc.h(qubit)
        qc.s(qubit)

    elif state_name == "arbitrary":
        qc.ry(1.1, qubit)
        qc.rz(0.7, qubit)

    else:
        raise ValueError("Unknown state")


def get_target_state(state_name):

    qc = QuantumCircuit(1)

    prepare_state(
        qc,
        0,
        state_name
    )

    return Statevector.from_instruction(qc)


# ============================================================
# 2. PAULI TWIRLING FUNCTIONS
# ============================================================

PAULIS = ["I", "X", "Y", "Z"]

pauli_to_bits = {
    "I": (0, 0),
    "X": (1, 0),
    "Y": (1, 1),
    "Z": (0, 1)
}

bits_to_pauli = {
    (0, 0): "I",
    (1, 0): "X",
    (1, 1): "Y",
    (0, 1): "Z"
}


def apply_pauli(qc, p, qubit):

    if p == "X":
        qc.x(qubit)

    elif p == "Y":
        qc.y(qubit)

    elif p == "Z":
        qc.z(qubit)


def cx_conjugated_paulis(pc, pt):

    xc, zc = pauli_to_bits[pc]
    xt, zt = pauli_to_bits[pt]

    return (
        bits_to_pauli[
            (
                xc,
                zc ^ zt
            )
        ],
        bits_to_pauli[
            (
                xt ^ xc,
                zt
            )
        ]
    )


def cz_conjugated_paulis(pc, pt):

    xc, zc = pauli_to_bits[pc]
    xt, zt = pauli_to_bits[pt]

    return (
        bits_to_pauli[
            (
                xc,
                zc ^ xt
            )
        ],
        bits_to_pauli[
            (
                xt,
                zt ^ xc
            )
        ]
    )


def twirled_cx(qc, control, target, rng):

    pc = rng.choice(PAULIS)
    pt = rng.choice(PAULIS)

    apply_pauli(qc, pc, control)
    apply_pauli(qc, pt, target)

    qc.cx(control, target)

    pc_after, pt_after = cx_conjugated_paulis(
        pc,
        pt
    )

    apply_pauli(
        qc,
        pc_after,
        control
    )

    apply_pauli(
        qc,
        pt_after,
        target
    )


def twirled_cz(qc, control, target, rng):

    pc = rng.choice(PAULIS)
    pt = rng.choice(PAULIS)

    apply_pauli(qc, pc, control)
    apply_pauli(qc, pt, target)

    qc.cz(control, target)

    pc_after, pt_after = cz_conjugated_paulis(
        pc,
        pt
    )

    apply_pauli(
        qc,
        pc_after,
        control
    )

    apply_pauli(
        qc,
        pt_after,
        target
    )


# ============================================================
# 3. NORMAL 5-QUBIT CIRCUIT
# ============================================================

def build_normal_5q(state_name):

    qc = QuantumCircuit(5)

    # Original state on q0
    prepare_state(
        qc,
        0,
        state_name
    )

    # ========================================================
    # TELEPORTATION 1:
    # q0 ---> q2
    # ========================================================

    # Bell pair q1-q2
    qc.h(1)
    qc.cx(1, 2)

    qc.barrier()

    # Bell-basis operation
    qc.cx(0, 1)
    qc.h(0)

    # Corrections onto q2
    qc.cx(1, 2)
    qc.cz(0, 2)

    qc.barrier()

    # q2 now contains the state


    # ========================================================
    # TELEPORTATION 2:
    # q2 ---> q4
    # ========================================================

    # Bell pair q3-q4
    qc.h(3)
    qc.cx(3, 4)

    qc.barrier()

    # Bell-basis operation q2-q3
    qc.cx(2, 3)
    qc.h(2)

    # Corrections onto q4
    qc.cx(3, 4)
    qc.cz(2, 4)

    qc.barrier()

    # q4 is final destination
    qc.save_density_matrix(
        qubits=[4],
        label="rho_out"
    )

    return qc


# ============================================================
# 4. TWIRLED 5-QUBIT CIRCUIT
# ============================================================

def build_twirled_5q(rng, state_name):

    qc = QuantumCircuit(5)

    prepare_state(
        qc,
        0,
        state_name
    )

    # ========================================================
    # TELEPORTATION 1
    # ========================================================

    qc.h(1)

    twirled_cx(
        qc,
        1,
        2,
        rng
    )

    qc.barrier()

    twirled_cx(
        qc,
        0,
        1,
        rng
    )

    qc.h(0)

    twirled_cx(
        qc,
        1,
        2,
        rng
    )

    twirled_cz(
        qc,
        0,
        2,
        rng
    )

    qc.barrier()

    # ========================================================
    # TELEPORTATION 2
    # ========================================================

    qc.h(3)

    twirled_cx(
        qc,
        3,
        4,
        rng
    )

    qc.barrier()

    twirled_cx(
        qc,
        2,
        3,
        rng
    )

    qc.h(2)

    twirled_cx(
        qc,
        3,
        4,
        rng
    )

    twirled_cz(
        qc,
        2,
        4,
        rng
    )

    qc.barrier()

    qc.save_density_matrix(
        qubits=[4],
        label="rho_out"
    )

    return qc


# ============================================================
# 5. COHERENT NOISE
# ============================================================

def make_coherent_noise_model(epsilon):

    noise_model = NoiseModel()

    error_circuit = QuantumCircuit(2)

    error_circuit.rx(
        epsilon,
        1
    )

    error_unitary = Operator(
        error_circuit
    ).data

    error = coherent_unitary_error(
        error_unitary
    )

    noise_model.add_all_qubit_quantum_error(
        error,
        ["cx", "cz"]
    )

    return noise_model


# ============================================================
# 6. FIDELITY
# ============================================================

def calculate_fidelity(
    circuit,
    noise_model,
    target_state
):

    simulator = AerSimulator(
        method="density_matrix",
        noise_model=noise_model
    )

    circuit = transpile(
        circuit,
        simulator,
        optimization_level=0,
        seed_transpiler=1234
    )

    result = simulator.run(
        circuit
    ).result()

    rho_out = result.data(0)[
        "rho_out"
    ]

    F = state_fidelity(
        target_state,
        rho_out
    )

    return np.clip(
        float(np.real(F)),
        0,
        1
    )


# ============================================================
# 7. ZERO-NOISE TEST
# ============================================================

print("5-QUBIT ZERO-NOISE CHECK\n")

no_noise = make_coherent_noise_model(
    0
)

for state_name in TEST_STATES:

    target = get_target_state(
        state_name
    )

    F = calculate_fidelity(
        build_normal_5q(
            state_name
        ),
        no_noise,
        target
    )

    print(
        f"{state_name}: {F:.10f}"
    )


# ============================================================
# 8. ACTUAL 5-QUBIT EXPERIMENT
# ============================================================

noise_strengths = np.linspace(
    0,
    0.30,
    13
)

N_TWIRLS = 100

rows = []


for state_name in TEST_STATES:

    print(
        f"\n===== 5Q STATE {state_name} ====="
    )

    target_state = get_target_state(
        state_name
    )

    for epsilon in noise_strengths:

        noise_model = make_coherent_noise_model(
            epsilon
        )

        # -------------------------------------
        # NO PT
        # -------------------------------------

        normal_circuit = build_normal_5q(
            state_name
        )

        F_normal = calculate_fidelity(
            normal_circuit,
            noise_model,
            target_state
        )

        # -------------------------------------
        # PT
        # -------------------------------------

        twirled_fidelities = []

        rng = np.random.default_rng(
            2026
        )

        for _ in range(N_TWIRLS):

            circuit = build_twirled_5q(
                rng,
                state_name
            )

            F = calculate_fidelity(
                circuit,
                noise_model,
                target_state
            )

            twirled_fidelities.append(
                F
            )

        PT_mean = np.mean(
            twirled_fidelities
        )

        PT_std = np.std(
            twirled_fidelities,
            ddof=1
        )

        PT_sem = PT_std / np.sqrt(
            N_TWIRLS
        )

        rows.append({
            "state": state_name,
            "epsilon": epsilon,
            "no_PT": F_normal,
            "PT_mean": PT_mean,
            "PT_std": PT_std,
            "PT_sem": PT_sem,
            "delta_F": PT_mean - F_normal
        })

        print(
            f"epsilon={epsilon:.3f} | "
            f"No PT={F_normal:.5f} | "
            f"PT={PT_mean:.5f}"
        )


# ============================================================
# 9. SAVE RESULTS
# ============================================================

results_5q = pd.DataFrame(rows)

print("\nCOMPLETE 5-QUBIT DATA:")
print(results_5q)

results_5q.to_csv(
    "5qubit_fidelity_results.csv",
    index=False
)


# ============================================================
# 10. GRAPH RESULTS
# ============================================================

for state_name in TEST_STATES:

    data = results_5q[
        results_5q["state"] == state_name
    ]

    plt.figure(figsize=(9, 6))

    plt.plot(
        data["epsilon"],
        data["no_PT"],
        marker="o",
        label="Without Pauli Twirling"
    )

    plt.errorbar(
        data["epsilon"],
        data["PT_mean"],
        yerr=data["PT_sem"],
        marker="o",
        capsize=4,
        label="With Pauli Twirling"
    )

    plt.xlabel(
        "Coherent Error Strength ε (radians)"
    )

    plt.ylabel(
        "Teleportation Fidelity"
    )

    plt.title(
        f"5-Qubit Sequential Teleportation — Input {state_name}"
    )

    plt.ylim(
        0,
        1.02
    )

    plt.grid(
        alpha=0.3
    )

    plt.legend()

    plt.tight_layout()

    plt.show()
```
